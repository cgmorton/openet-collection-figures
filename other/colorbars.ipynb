```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

```

```python
min_val = 0
max_val = 12

# Create some dummy data and a colormap for the colorbar
fig, ax = plt.subplots(figsize=(3, 0.5)) # Adjust figsize for desired dimensions
fig.subplots_adjust(bottom=0.5) # Adjust spacing if needed

ticks = list(range(0, max_val+1))
boundaries = np.linspace(min_val-0.5, max_val+0.5, max_val-min_val+2)
cmap = plt.get_cmap('viridis')
norm = mcolors.BoundaryNorm(boundaries, cmap.N)
cbar = fig.colorbar(
    plt.cm.ScalarMappable(norm=norm, cmap=cmap),
    cax=ax,
    orientation='horizontal',
    label='Monthly Image Count',
    ticks=ticks,
)
cbar.ax.minorticks_off()

fig.savefig('horizontal_colorbar.png', dpi=300, bbox_inches='tight')
plt.show()
```
