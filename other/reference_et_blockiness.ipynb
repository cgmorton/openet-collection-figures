```python
import pprint

import ee
from IPython.display import Image, display
#import ipyplot
#import matplotlib.pyplot as plt
#import numpy as np
#import openet.core
```

```python
ee.Initialize()
```

```python
start_date = '2024-07-01'
end_date = '2024-08-01'

#region = ee.Geometry.BBox(-125.0, 35, -120.0, 42.0)
#region = ee.Geometry.BBox(-124.0, 38.5, -122.0, 39.9)
region = ee.Geometry.BBox(-123.9, 38.6, -122.7, 39.8)

image_size = 300

et_palette = ['DEC29B', 'E6CDA1', 'EDD9A6', 'F5E4A9', 'FFF4AD', 'C3E683', '6BCC5C', '3BB369', '20998F', '1C8691', '16678A', '114982', '0B2C7A']
viridis = ['440154', '433982', '30678D', '218F8B', '36B677', '8ED542', 'FDE725']
```

```python
model = 'disalexi'
version = 'v2_0'
image_url = (
    ee.ImageCollection(f'projects/openet/assets/{model}/conus/gridmet/monthly/{version}')
    .filterDate(start_date, end_date).select(['et']).sum()
    .visualize(min=0, max=300, palette=et_palette).getThumbURL({'region': region, 'dimensions': image_size})
)
Image(image_url, embed=True, format='png')  
```
